```python
#1.Create a Pandas DataFrame with at least 5 columns representing different features of food delivery orders (e.g., delivery_time, order_value, 
#tip_amount, 
#rating, delivery_distance) and 20 rows of random data, then calculate and display the correlation matrix using the .corr() function.

import numpy as np
import pandas as pd

# Create a mock food delivery DataFrame with 5 columns and 20 rows
np.random.seed(42)

data = {
    "delivery_time_min": np.random.randint(15, 60, size=20),
    "order_value_inr": np.random.randint(150, 1200, size=20),
    "tip_amount_inr": np.random.randint(10, 150, size=20),
    "rating_stars": np.random.uniform(2.5, 5.0, size=20).round(1),
    "delivery_distance_km": np.random.uniform(1.0, 12.0, size=20).round(1),
}

df_orders = pd.DataFrame(data)

# Calculate the correlation matrix using .corr()
# By default, this uses Pearson correlation to check linear relationships
correlation_matrix = df_orders.corr()

# Display the mathematical matrix
print("--- Food Delivery Order Correlation Matrix ---")
print(correlation_matrix.round(2))
```

```python
#2.Using Seaborn, plot a heatmap of the correlation matrix you 
#generated for the food delivery data, and add annotations to show the correlation values inside each cell.

import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the correlation matrix
corr_matrix = df_orders.corr()

# Set up the plotting window and style
sns.set_theme(style="white")
plt.figure(figsize=(8, 6))

# Draw the heatmap with cell annotations
# annot=True prints the mathematical values inside each grid square
# fmt=".2f" ensures the numbers are clean and rounded to 2 decimal places
sns.heatmap(data=corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)

# Add a title
plt.title("Correlation Matrix Heatmap: Food Delivery Metrics", fontsize=14, fontweight="bold", pad=15)

plt.tight_layout()
plt.show()
```

```python
#3.Change the color palette of your Seaborn heatmap to a diverging palette (such as 'coolwarm' or 'RdYlGn') to better visualize positive and 
#negative correlations, and explain 
#which two features in your data have the strongest negative correlation.<br><br><em><strong>Hint:</strong> Use the cmap parameter in sns.heatmap().</em>

#Calculate the correlation matrix
corr_matrix = df_orders.corr()

# Set up the plotting window
sns.set_theme(style="white")
plt.figure(figsize=(8, 6))

# Draw the heatmap with the 'RdYlGn' diverging palette
# vmin and vmax set the absolute boundaries of the correlation scale (-1 to +1)
sns.heatmap(data=corr_matrix,annot=True,fmt=".2f",cmap="coolwarm",vmin=-1,vmax=1,linewidths=0.5)

# Add a title
plt.title("Diverging Correlation Heatmap (Coolwarm Palette)",fontsize=14,fontweight="bold",pad=15)

# Display the window
plt.tight_layout()
plt.show()
```

```python
#4.Take the Flipkart mobile phones dataset (or any small CSV with product features you can find online), compute the correlation matrix for 
#at least 4 numeric columns (e.g., price, rating, RAM, battery_capacity), and generate a Seaborn heatmap with annotations and a diverging color palette.

#Create a sample Flipkart mobile dataset so the code runs directly ---
np.random.seed(101)
mobile_data = {
    "price_inr": np.random.randint(9999, 89999, size=30),
    "rating_stars": np.random.uniform(3.8, 4.9, size=30).round(1),
    "ram_gb": np.random.choice([4, 6, 8, 12, 16], size=30),
    "battery_capacity_mah": np.random.choice([4000,4500,5000,6000],size=30)
}
pd.DataFrame(mobile_data).to_csv("flipkart_mobiles.csv", index=False)


# Load the CSV file and compute the correlation matrix ---
df_mobiles = pd.read_csv("flipkart_mobiles.csv")
mobile_corr = df_mobiles.corr()

# Setup and plot the diverging heatmap 
sns.set_theme(style="white")
plt.figure(figsize=(8, 6))

# Center=0 anchors the middle point of the colorbar at zero (neutral color)
sns.heatmap(data=mobile_corr,annot=True,fmt=".2f",cmap="coolwarm",vmin=-1,vmax=1,center=0,linewidths=0.8)

# Add labels and titles
plt.title("Flipkart Mobile Phone Feature Correlations",fontsize=14,fontweight="bold",pad=15)
plt.tight_layout()
plt.show()
```
