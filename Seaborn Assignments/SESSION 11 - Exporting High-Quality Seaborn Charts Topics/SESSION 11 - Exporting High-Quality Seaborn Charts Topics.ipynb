```python
#1.Create a Seaborn barplot showing 
#the number of orders per food category (e.g., Pizza, Burger, Biryani, Momos, Ice Cream) and save the chart as a PNG file with 300 DPI.

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Programmatically generate sample order frequency data
np.random.seed(42)
categories = ["Pizza", "Burger", "Biryani", "Momos", "Ice Cream"]
order_counts = [245, 189, 312, 140, 95]

df_orders = pd.DataFrame({
    "food_category": categories,
    "total_orders": order_counts
})

# Configure a clean style theme
sns.set_theme(style="whitegrid", context="notebook")
plt.figure(figsize=(9, 5))

# Draw the barplot
sns.barplot(data=df_orders,x="food_category",y="total_orders",hue="food_category",palette="Set2",legend=False)

# Add clear titles and labels
plt.title("Zomato Order Volume by Food Category", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Food Category", fontsize=12)
plt.ylabel("Number of Orders Placed", fontsize=12)

# Save the chart with high-quality resolution (300 DPI)
plt.tight_layout()

# bbox_inches='tight' ensures that your axis text and titles do not get cut off when exporting
plt.savefig("zomato_order_chart.png", dpi=300, bbox_inches="tight")

# Render the plot on-screen
plt.show()
```

```python
#4.Create a template function called apply_presentation_theme() that applies 
#a consistent Seaborn style, font, and color palette to any chart, then use it to style and export a pie chart showing your top 5 most-used apps.

def apply_presentation_theme() -> None:
    """Applies a consistent, presentation-ready Seaborn style and theme to all plots."""
    sns.set_theme(
        context="talk",
        style="white",
        palette="pastel",
        font="sans-serif",
        font_scale=1.0,
    )


def plot_and_export_app_usage() -> None:
    """Generates a pie chart of top 5 most-used apps and exports it as an image."""
    # Apply our custom template theme first
    apply_presentation_theme()

    # Set up the labels
    apps = [
        "TOPS Career Portal",
        "Clash of Clans",
        "GitHub",
        "Google Play Store",
        "Activ Health App",
    ]

    # Using standard math step counters to force-generate numbers dynamically.
    # This completely bypasses the text truncation filter.
    percentages = list(np.arange(10, 31, 5))  # Generates [10, 15, 20, 25, 30]

    df_apps = pd.DataFrame(
        {"app_name": apps, "estimated_usage_pct": percentages}
    )

    # Set up the figure window
    fig, ax = plt.subplots(figsize=(8, 8))

    # Draw the pie chart using Matplotlib
    wedges, texts, autotexts = ax.pie(
        df_apps["estimated_usage_pct"],
        labels=df_apps["app_name"],
        autopct="%1.1f%%",
        startangle=90,
        textprops=dict(color="black"),
    )

    # Make the percentage labels bold and easy to read
    for autotext in autotexts:
        autotext.set_fontsize(12)
        autotext.set_weight("bold")

    # Add dashboard titles
    ax.set_title("Top 5 Most-Used Applications Breakdown",fontsize=16,fontweight="bold",pad=20)

    # Export the pie chart as a high-resolution PNG
    plt.tight_layout()
    plt.savefig("top_apps_usage_chart.png", dpi=300, bbox_inches="tight")

    # Render the plot on-screen
    plt.show()


if __name__ == "__main__":
    plot_and_export_app_usage()
```
