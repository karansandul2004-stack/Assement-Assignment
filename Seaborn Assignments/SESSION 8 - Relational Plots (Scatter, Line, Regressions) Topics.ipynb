```python
#1.Use sns.scatterplot() to visualize 
#the relationship between the number of followers and the number of posts for 20 fictional Instagram users. Label the axes appropriately.

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Create a fictional Instagram dataset for 20 users
np.random.seed(42)
instagram_data = {
    # Number of posts ranges from 20 to 500
    "number_of_posts": np.random.randint(20, 501, size=20),
    # Follower count roughly scales with posts, adding some random variance
    "follower_count": np.random.randint(500, 50000, size=20),
}
df_instagram = pd.DataFrame(instagram_data)

# Set the visual theme and figure window
sns.set_theme(style="ticks")
plt.figure(figsize=(8, 5))

# Draw the scatter plot
# alpha=0.8 makes the dots slightly transparent so overlaps are easier to see
sns.scatterplot(data=df_instagram,x="number_of_posts",y="follower_count",color="darkmagenta",s=100,alpha=0.8)
# Sets the size of the scatter dots

# Add appropriate student-friendly labels and titles
plt.title("Instagram Engagement: Posts vs. Follower Count",fontsize=14,fontweight="bold",pad=15)
plt.xlabel("Number of Posts", fontsize=12)
plt.ylabel("Follower Count", fontsize=12)

# Clean layout boundaries and render the plot
plt.tight_layout()
plt.show()
```

```python
#2.Create a line plot using sns.lineplot() to show 
#the daily number of food orders on Zomato for each day over a month. Use a list of 30 random order counts to simulate the data.

#Create simulated data for 30 days of Zomato orders
np.random.seed(55)
zomato_data = {
    # Sequential days from 1 to 30
    "day_of_month": np.arange(1, 31),
    # Random daily order counts between 150 and 450 orders
    "daily_orders": np.random.randint(150, 451, size=30),
}
df_zomato = pd.DataFrame(zomato_data)

# Set the visual theme and size
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 5))

# marker='o' adds explicit structural dots onto the line for each data day
sns.lineplot(data=df_zomato,x="day_of_month",y="daily_orders",color="crimson",linewidth=2.5,marker="o")

# Add clear titles and step metrics
plt.title("Zomato Daily Order Trends Over One Month",fontsize=14,fontweight="bold",pad=15,)
plt.xlabel("Day of the Month", fontsize=12)
plt.ylabel("Number of Orders", fontsize=12)

# Force the X-axis to display every few days cleanly
plt.xticks(np.arange(1, 31, step=2))

# Render the plot
plt.tight_layout()
plt.show()
```

```python
#3.Using sns.relplot(), plot the relationship between movie ratings and the number of reviews for different genres (Action, Comedy, Drama) based on a 
#sample dataset of 15 movies. Use the 
#'hue' parameter to differentiate genres.<br><br><em><strong>Hint:</strong> Create a DataFrame with columns: 'movie', 'rating', 'reviews'
#, and 'genre'.</em>

#Create a custom dataset of exactly 15 fictional movies
# Following the hint, we include movie name, rating, reviews, and genre
movie_data = {
    "movie": [
        "Action Hero 1", "Action Hero 2", "Super Speedway", "Cyber Hunter", "Urban Legend",
        "Laughter Club", "Office Pranks", "Standup Star", "Crazy Weekend", "Comic Relief",
        "Family Ties", "Silent Tears", "Hustle Life", "Autumn Leaves", "The Crossroads"
    ],
    "rating": [4.2, 3.8, 4.5, 3.5, 4.0, 4.1, 4.7, 3.9, 4.3, 3.6, 4.4, 4.6, 3.7, 4.2, 4.0],
    "reviews": [120, 85, 340, 95, 150, 450, 620, 310, 280, 190, 800, 950, 210, 540, 410],
    "genre": [
        "Action", "Action", "Action", "Action", "Action",
        "Comedy", "Comedy", "Comedy", "Comedy", "Comedy",
        "Drama", "Drama", "Drama", "Drama", "Drama"
    ]
}
df_movies = pd.DataFrame(movie_data)

# Set the style grid
sns.set_theme(style="ticks")

# Draw the chart using relplot (figure-level plot)
# kind="scatter" is used by default, but we declare it to be clear
g = sns.relplot(data=df_movies,x="rating",y="reviews",hue="genre",kind="scatter",palette="Set1",s=120,alpha=0.9)  # Clear marker sizing for 15 elements

# Add clear titles and labels using the relplot facet layout
g.fig.suptitle("Movie Performance: Ratings vs. Review Volume", fontsize=14, fontweight="bold", y=1.05)
g.set_axis_labels("User Rating (out of 5)", "Number of Reviews", fontsize=12)

# Render the plot window
plt.show()
```

```python
#4.Generate a regression plot using sns.regplot() to analyze the correlation between 
#the number of hours spent on YouTube and exam scores for a group of 25 students. Interpret whether the trend is positive or negative.

# Programmatically generate realistic data for 25 students
np.random.seed(42)

# Generate study/watch hours between 0.5 and 6 hours per day
youtube_hours = np.random.uniform(0.5, 6.0, size=25)

# Create exam scores with a negative relationship to screen time
# Base score of 95 drops by roughly 7 marks per hour watched, with some random variance
exam_scores = 95 - (youtube_hours * 7) + np.random.normal(loc=0, scale=4, size=25)

# Clip the scores so they stay within a valid 0 to 100 percentage range
exam_scores = np.clip(exam_scores, 0, 100)

student_df = pd.DataFrame({
    "hours_on_youtube": youtube_hours,
    "exam_score": exam_scores
})

# Set the style grid
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))

# Draw the regression plot
# sns.regplot automatically plots the scatter dots AND draws the best-fit line
sns.regplot(data=student_df,x="hours_on_youtube",y="exam_score",color="crimson",marker="x",scatter_kws={"s": 80})  
# Makes the individual student 'x' marks larger

# Add clear titles and labels
plt.title("Impact of YouTube Screen Time on Student Exam Scores", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Daily Hours Spent on YouTube", fontsize=12)
plt.ylabel("Exam Score (%)", fontsize=12)

plt.tight_layout()
plt.show()
```
