```python
#1.Use sns.set() to apply the 'poster' context to a Seaborn barplot showing the number of 
#songs streamed per day on Spotify, and observe how the plot elements change compared to the default context.

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Programmatically generate sample Spotify streaming data for a week
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
np.random.seed(42)
spotify_data = {
    "day_of_week": days,
    "songs_streamed": np.random.randint(15, 45, size=7)
}
df_spotify = pd.DataFrame(spotify_data)

# Set the scaling context to 'poster' using sns.set_theme()
# (Note: In modern Seaborn versions, sns.set_theme() is used in place of sns.set())
sns.set_theme(context="poster", style="whitegrid")

plt.figure(figsize=(10, 6))
sns.barplot(data=df_spotify,x="day_of_week",y="songs_streamed",hue="day_of_week",palette="deep",legend=False)

# Add clear titles and labels
plt.title("Daily Spotify Streams", pad=20, fontweight="bold")
plt.xlabel("Day of Week")
plt.ylabel("Songs Streamed")

# Render the giant-styled plot
plt.tight_layout()
plt.show()
```

```python
#2.Create a Seaborn countplot showing the distribution of food orders by cuisine type (e.g., Indian, Chinese, Italian) using a 'colorblind' palette, 
#similar to how Zomato might visualize order trends.<br><br><em><strong>Hint:</strong> Use palette='colorblind' in your plot function.</em>

# Programmatically generate sample Zomato order data for 150 entries
np.random.seed(42)
cuisines = ["Indian", "Chinese", "Italian", "Mexican", "Continental"]

df_zomato = pd.DataFrame({
    "cuisine_type": np.random.choice(cuisines, size=150, p=[0.4, 0.25, 0.15, 0.1, 0.1])
})

# Set the style configuration back to 'notebook' for standard clean layouts
sns.set_theme(context="notebook", style="ticks")
plt.figure(figsize=(9, 5))

# Draw the countplot using the 'colorblind' palette constraint
sns.countplot(data=df_zomato,x="cuisine_type",hue="cuisine_type",palette="colorblind",legend=False,order=cuisines)  
# Keeps the bars in a clean, predictable order

# Add student-friendly labels and titles
plt.title("Zomato Cuisine Order Trends (Accessible Color Blind View)", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Cuisine Type", fontsize=12)
plt.ylabel("Number of Orders Placed", fontsize=12)

# Clean up borders and render the chart
sns.despine()  # Removes top and right spine lines for a modern look
plt.tight_layout()
plt.show()
```

```python
#3.Generate a Seaborn lineplot of daily step counts 
#(simulate data for 2 weeks) and style it using the 'muted' palette and 'notebook' context for a clean, presentation-ready look.

# Programmatically generate 2 weeks of fitness data for two sample users
np.random.seed(24)
days_range = np.arange(1, 15)

# Simulate step data for User A and User B
steps_user_a = np.random.normal(loc=8500, scale=1200, size=14).astype(int)
steps_user_b = np.random.normal(loc=7200, scale=900, size=14).astype(int)

df_fitness = pd.DataFrame({
    "day": np.concatenate([days_range, days_range]),
    "daily_steps": np.concatenate([steps_user_a, steps_user_b]),
    "user_profile": ["User A"] * 14 + ["User B"] * 14
})

# Establish presentation-ready styling with 'notebook' context
# style="white" drops heavy grid backgrounds for a highly professional feel
sns.set_theme(context="notebook", style="white")
plt.figure(figsize=(10, 5))

# Draw the line plot with the 'muted' palette configuration
# marker="o" anchors specific days, and markersize coordinates with notebook scaling
sns.lineplot(data=df_fitness,x="day",y="daily_steps",hue="user_profile",palette="muted",linewidth=2.5,marker="o",markersize=7)

# Add clear titles, labels, and ticks
plt.title("Google Fit Tracker: Daily Step Trends (2-Week Summary)", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Day of Tracking", fontsize=12)
plt.ylabel("Total Steps Taken", fontsize=12)

# Ensure the x-axis displays clean day counts from 1 to 14
plt.xticks(days_range)

# Professional frame polish and rendering
sns.despine() # Strip top and right frame lines
plt.tight_layout()
plt.show()
```

```python
#4.Refactor an existing Seaborn plot by switching its palette from 'deep' to 'colorblind' 
#and changing the context from 'talk' to 'poster'. Briefly describe the visual differences you notice after applying these changes.

#Programmatically generate sample order status data
np.random.seed(42)
statuses = ["Preparing", "Dispatched", "Out for Delivery", "Delivered"]
df_orders = pd.DataFrame({
    "order_status": np.random.choice(statuses, size=100, p=[0.1, 0.2, 0.3, 0.4])
})

# --- CONFIGURATION 1: BEFORE (deep + talk) ---
# We use rc parameters manually to mimic the 'talk' context inside subplots
sns.set_theme(context="talk", style="ticks")
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

sns.countplot(data=df_orders,x="order_status",hue="order_status",palette="deep",legend=False,order=statuses,ax=axes[0])
axes[0].set_title("Before: 'deep' Palette + 'talk' Context", fontsize=16, fontweight="bold")
axes[0].tick_params(axis='x', rotation=15)

# --- CONFIGURATION 2: AFTER (colorblind + poster) ---
# We update the global parameters to 'poster' context for the second plot
sns.set_theme(context="poster", style="ticks")

sns.countplot(data=df_orders,x="order_status",hue="order_status",palette="colorblind",legend=False,order=statuses,ax=axes[1])
axes[1].set_title("After: 'colorblind' Palette + 'poster' Context", fontweight="bold")
axes[1].tick_params(axis='x', rotation=15)

# Clear extra line borders and render
sns.despine()
plt.tight_layout()
plt.show()
```

```python
#5.Design a Seaborn theme that mimics the look of a Flipkart sales dashboard: use 
#sns.set() to set a 'deep' palette, 'talk' context, and customize the font scale and grid style. Apply your theme to a barplot showing sales of 
#5 product categories.

#Programmatically generate sample Flipkart sales data for 5 categories
np.random.seed(99)
categories = ["Electronics", "Fashion", "Home & Kitchen", "Beauty & Personal Care", "Books & Stationery"]
sales_revenue_lakhs = [45.2, 38.7, 24.5, 18.9, 12.4]

df_sales = pd.DataFrame({
    "product_category": categories,
    "sales_revenue": sales_revenue_lakhs
})

# Design and apply the custom Flipkart-style dashboard theme
# We set style='whitegrid' for subtle horizontal tracking lines, 
# context='talk' for clear presentation scaling, and increase font_scale to 1.1
sns.set_theme(context="talk",style="whitegrid",palette="deep",font_scale=1.1,
    rc={
        "grid.linestyle": "--",     # Changes solid grid lines to clean dashed lines
        "grid.linewidth": 0.8,      # thins the grid lines so they don't distract
        "axes.edgecolor": "#CCCCCC" # Softens the outer border lines
    }
)

# Draw the dashboard barplot
plt.figure(figsize=(11, 6))
sns.barplot(data=df_sales,x="product_category",y="sales_revenue",hue="product_category",legend=False)

# Add clear dashboard titles and labels
plt.title("Flipkart Festive Sale: Revenue Breakdown by Category", fontsize=16, fontweight="bold", pad=20, color="#111111")
plt.xlabel("Product Category", fontsize=13, fontweight="bold", labelpad=12)
plt.ylabel("Sales Revenue (in Lakhs ₹)", fontsize=13, fontweight="bold", labelpad=12)

# Tilt the category names slightly so they don't overlap under the talk scale
plt.xticks(rotation=15)

# Render the final dashboard view
plt.tight_layout()
plt.show()
```
