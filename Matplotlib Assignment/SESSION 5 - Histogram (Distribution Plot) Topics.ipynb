```python
#1.Use plt.hist() to plot a histogram of delivery times (in minutes) for 30 Swiggy orders: 
#[32, 28, 45, 30, 36, 40, 29, 34, 38, 31, 27, 41, 33, 35, 39, 37, 44, 30, 29, 32, 36, 28, 43, 31, 35, 40, 38, 33, 30, 42]. Show the plot window.

import matplotlib.pyplot as plt

# Prepare data
delivery_times = [32, 28, 45, 30, 36, 40, 29, 34, 38, 31, 
                  27, 41, 33, 35, 39, 37, 44, 30, 29, 32, 
                  36, 28, 43, 31, 35, 40, 38, 33, 30, 42]

# Create histogram
plt.hist(delivery_times, bins=6, color="skyblue", edgecolor="black")

# Add labels and title
plt.title("Histogram of Swiggy Delivery Times")
plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Frequency")

plt.show()
```

```python
#2.Plot a histogram of IPL match scores (runs per team in a match) using plt.hist() and experiment with bins=5, bins=10, and bins=15. 
#Observe and describe how changing 
#the number of bins affects the shape of the histogram.<br><br><em><strong>Hint:</strong> Use any sample array of 50-60 scores between 120 and 220.</em>

import numpy as np

# Generate sample data (50 scores between 120 and 220)
np.random.seed(42)  # for reproducibility
scores = np.random.randint(120, 221, size=55)

# Plot histogram with bins=5
plt.hist(scores, bins=5, color="skyblue", edgecolor="black")
plt.title("IPL Match Scores Distribution (bins=5)")
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.show()

# Plot histogram with bins=10
plt.hist(scores, bins=10, color="lightgreen", edgecolor="black")
plt.title("IPL Match Scores Distribution (bins=10)")
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.show()

# Plot histogram with bins=15
plt.hist(scores, bins=15, color="salmon", edgecolor="black")
plt.title("IPL Match Scores Distribution (bins=15)")
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.show()
```

```python
#3.Given an array of daily step counts from a fitness tracker app for one month, plot 
#two histograms using plt.hist(): one showing frequency (default), and one showing density=True. Compare the y-axis labels and explain the difference.

# Sample data (30 daily step counts)
np.random.seed(42)  # reproducibility
steps = np.random.randint(3000, 15000, size=30)

# Frequency histogram (default)
plt.hist(steps, bins=8, color="skyblue", edgecolor="black")
plt.title("Daily Step Counts - Frequency")
plt.xlabel("Steps")
plt.ylabel("Frequency")
plt.show()

# Density histogram
plt.hist(steps, bins=8, color="lightgreen", edgecolor="black", density=True)
plt.title("Daily Step Counts - Density")
plt.xlabel("Steps")
plt.ylabel("Probability Density")
plt.show()
```

```python
#4.Simulate the distribution of song durations (in seconds) from your Spotify playlist by generating 50 
#random numbers between 120 and 300. Plot a histogram using plt.hist() and set the color to 'purple' and add the title 'Spotify Song Duration Distribution'.

# Generate random song durations (50 songs, 120–300 seconds)
np.random.seed(42)  # reproducibility
durations = np.random.randint(120, 301, size=50)

# Create histogram
plt.hist(durations, bins=8, color="purple", edgecolor="black")

# Add title and labels
plt.title("Spotify Song Duration Distribution")
plt.xlabel("Duration (seconds)")
plt.ylabel("Frequency")

plt.show()
```
