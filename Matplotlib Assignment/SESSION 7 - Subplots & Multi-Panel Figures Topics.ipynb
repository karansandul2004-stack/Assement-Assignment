```python
#1.Use plt.subplot() to create a figure with 2 
#rows and 1 column of plots showing your daily step count and water intake (in litres) for the last 7 days. Plot both as line charts, one in each subplot.

import matplotlib.pyplot as plt

# Prepare data (last 7 days)
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [6500, 7200, 8000, 7500, 9000, 10000, 8500]
water = [2.5, 2.8, 3.0, 2.7, 3.2, 3.5, 3.0]  # litres

# Create subplots (2 rows, 1 column)
plt.figure(figsize=(8,6))

plt.subplot(2,1,1)
plt.plot(days, steps, marker="o", color="green")
plt.title("Daily Step Count (Last 7 Days)")
plt.ylabel("Steps")

# Water Intake
plt.subplot(2,1,2)
plt.plot(days, water, marker="o", color="blue")
plt.title("Daily Water Intake (Last 7 Days)")
plt.xlabel("Day")
plt.ylabel("Litres")

# Adjust layout and show
plt.tight_layout()
plt.show()
```

```python
#2.Create a 2x2 multi-panel figure using plt.subplots(). In each subplot, display the number of 
#Instagram posts you made in each quarter (Q1, Q2, Q3, Q4) for the past year, using different chart types: line, bar, scatter, and pie.

quarters = ["Q1", "Q2", "Q3", "Q4"]
posts = [15, 25, 20, 30]  # example data

# Create 2x2 subplots
fig, axs = plt.subplots(2, 2, figsize=(10,8))

# Line chart
axs[0,0].plot(quarters, posts, marker="o", color="blue")
axs[0,0].set_title("Line Chart")
axs[0,0].set_ylabel("Posts")

# Bar chart
axs[0,1].bar(quarters, posts, color="green")
axs[0,1].set_title("Bar Chart")

# Scatter plot
axs[1,0].scatter(quarters, posts, color="red", marker="o")
axs[1,0].set_title("Scatter Plot")
axs[1,0].set_ylabel("Posts")
axs[1,0].set_xlabel("Quarter")

# Pie chart
axs[1,1].pie(posts, labels=quarters, autopct="%1.1f%%", 
             colors=["orange","purple","cyan","pink"], startangle=90)
axs[1,1].set_title("Pie Chart")

# Adjust layout and show
plt.suptitle("Instagram Posts per Quarter (Different Chart Types)", fontsize=14)
plt.tight_layout(rect=[0,0,1,0.95])
plt.show()
```

```python
#3.Make a grid of 1 row and 3 columns using plt.subplots(), and plot the number of food orders you placed on Zomato, Swiggy, and Domino's over the last 6 months,
#one app per subplot. Share the y-axis 
#among all subplots so you can easily compare order counts across apps.<br><br><em><strong>Hint:</strong> Use the 'sharey' parameter in plt.subplots().</em>

# Prepare data (last 6 months)
months = ["May", "Jun", "Jul", "Aug", "Sep", "Oct"]
zomato_orders = [25, 30, 28, 35, 40, 38]
swiggy_orders = [20, 27, 32, 30, 38, 36]
dominos_orders = [15, 18, 20, 22, 25, 23]

# Create 1x3 subplots with shared y-axis
fig, axs = plt.subplots(1, 3, figsize=(12,4), sharey=True)

# Zomato
axs[0].plot(months, zomato_orders, marker="o", color="red")
axs[0].set_title("Zomato Orders")
axs[0].set_ylabel("Number of Orders")

# Swiggy
axs[1].plot(months, swiggy_orders, marker="s", color="blue")
axs[1].set_title("Swiggy Orders")

# Domino's
axs[2].plot(months, dominos_orders, marker="^", color="green")
axs[2].set_title("Domino's Orders")

# Add overall title and adjust layout
plt.suptitle("Food Orders Over Last 6 Months", fontsize=14)
plt.tight_layout(rect=[0,0,1,0.95])
plt.show()
```

```python
#4.Arrange a 2x2 grid of subplots using plt.subplots() to visualize your monthly mobile data usage (in GB) for four different apps: 
#YouTube, Instagram, Spotify,
#and WhatsApp. Ensure all 
#subplots share the same x-axis (months) for easy comparison.<br><br><em><strong>Hint:</strong> Use the 'sharex' parameter in plt.subplots().</em>

import matplotlib.pyplot as plt

# Prepare data (last 6 months)
months = ["May", "Jun", "Jul", "Aug", "Sep", "Oct"]
youtube_data = [12, 15, 14, 18, 20, 17]
instagram_data = [8, 10, 9, 11, 13, 12]
spotify_data = [6, 7, 6, 8, 9, 7]
whatsapp_data = [4, 5, 4, 6, 7, 5]

# Create 2x2 subplots with shared x-axis
fig, axs = plt.subplots(2, 2, figsize=(10,8), sharex=True)

# Subplot 1: YouTube
axs[0,0].plot(months, youtube_data, marker="o", color="red")
axs[0,0].set_title("YouTube Data Usage")
axs[0,0].set_ylabel("GB")

# Subplot 2: Instagram
axs[0,1].plot(months, instagram_data, marker="s", color="blue")
axs[0,1].set_title("Instagram Data Usage")

# Subplot 3: Spotify
axs[1,0].plot(months, spotify_data, marker="^", color="green")
axs[1,0].set_title("Spotify Data Usage")
axs[1,0].set_ylabel("GB")
axs[1,0].set_xlabel("Month")

# Subplot 4: WhatsApp
axs[1,1].plot(months, whatsapp_data, marker="d", color="purple")
axs[1,1].set_title("WhatsApp Data Usage")
axs[1,1].set_xlabel("Month")

# Add overall title and adjust layout
plt.suptitle("Monthly Mobile Data Usage by App", fontsize=14)
plt.tight_layout(rect=[0,0,1,0.95])
plt.show()
```
