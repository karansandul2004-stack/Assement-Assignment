```python
#1.Use plt.bar() to create a vertical bar chart showing the number of daily orders for 5 popular food delivery apps 
#(e.g., Zomato, Swiggy, Uber Eats, Dunzo, Domino's) and set a custom color for each bar.

import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [120, 95, 80, 60, 100]  # made-up data

# Define custom colors for each bar
colors = ["red", "blue", "green", "orange", "purple"]

# Create vertical bar chart
plt.bar(apps, orders, color=colors)

# Add labels and title
plt.title("Daily Orders by Food Delivery App")
plt.xlabel("App")
plt.ylabel("Number of Orders")

plt.show()
```

```python
#2.Create a horizontal bar chart using plt.barh() to display 
#the number of followers for 6 different Instagram influencers; add a title and label the x-axis as 'Followers (in millions)'.

influencers = ["Influencer A", "Influencer B", "Influencer C", 
               "Influencer D", "Influencer E", "Influencer F"]
followers = [12, 18, 25, 20, 15, 22]  # in millions (example data)

# Create horizontal bar chart
plt.barh(influencers, followers, color="skyblue")

# Add title and axis labels
plt.title("Instagram Influencers - Followers Count")
plt.xlabel("Followers (in millions)")

plt.show()
```

```python
#3.Modify your vertical bar chart to display the exact order count as a text label above each bar using 
#plt.text().<br><br><em><strong>Hint:</strong> Use a for loop to iterate through the bars and place the text labels at the correct height.</em>

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [120, 95, 80, 60, 100]  # made-up data
colors = ["red", "blue", "green", "orange", "purple"]

# Create vertical bar chart
plt.bar(apps, orders, color=colors)

# Add labels and title
plt.title("Daily Orders by Food Delivery App")
plt.xlabel("App")
plt.ylabel("Number of Orders")

# Add text labels above each bar
for i, value in enumerate(orders):
    plt.text(i, value + 2, str(value), ha="center", va="bottom", fontsize=10)

plt.show()
```

```python
#4.Change the width of the bars in your food delivery app chart to make them thinner, and adjust the spacing so 
#the bars do not overlap.<br><br><em><strong>Constraint:</strong> Use the 'width' parameter in plt.bar() and experiment with values less than 0.8.</em>

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [120, 95, 80, 60, 100]
colors = ["red", "blue", "green", "orange", "purple"]

# Create vertical bar chart with thinner bars
plt.bar(apps, orders, color=colors, width=0.5)  # width < 0.8 makes bars thinner

# Add labels and title
plt.title("Daily Orders by Food Delivery App")
plt.xlabel("App")
plt.ylabel("Number of Orders")

# Add text labels above each bar
for i, value in enumerate(orders):
    plt.text(i, value + 2, str(value), ha="center", va="bottom", fontsize=10)

plt.show()
```
