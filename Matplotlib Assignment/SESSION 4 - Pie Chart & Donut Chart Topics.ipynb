```python
#1.Use plt.pie() to visualize the percentage of time you spend on 5 different apps in a day 
#(e.g., Instagram, YouTube, WhatsApp, Zomato, Spotify) and display the percentage values on each segment using the autopct parameter.

import matplotlib.pyplot as plt

# Prepare data
apps = ["Instagram", "YouTube", "WhatsApp", "Zomato", "Spotify"]
time_spent = [120, 90, 60, 30, 100]  # minutes spent (example data)

#Create pie chart
plt.pie(time_spent, labels=apps, autopct="%1.1f%%", 
        colors=["pink", "red", "green", "orange", "blue"], 
        startangle=90)

# Add title
plt.title("Daily Time Spent on Apps")

plt.show()
```

```python
#2.Create a pie chart showing the distribution of your monthly online spending across 4 categories: Food Delivery, Shopping, Entertainment, and UPI Payments,
#and use the explode parameter to highlight the 
#category with the highest spending.<br><br><em><strong>Hint:</strong> Set a higher explode value for the segment you want to pull out from the pie.</em>

categories = ["Food Delivery", "Shopping", "Entertainment", "UPI Payments"]
spending = [5000, 12000, 8000, 6000]  # example values in INR

# Find the highest spending category
max_index = spending.index(max(spending))

# Create explode list (0 for others, 0.1 for highest)
explode = [0.1 if i == max_index else 0 for i in range(len(spending))]

# Create pie chart
plt.pie(spending, labels=categories, autopct="%1.1f%%", 
        explode=explode, startangle=90, 
        colors=["orange", "blue", "green", "purple"])

plt.title("Monthly Online Spending Distribution")

plt.show()
```

```python
#3.Transform your pie chart into a donut chart to show the same spending data by adding a white circle at the center using plt.Circle and ax.add_artist.

categories = ["Food Delivery", "Shopping", "Entertainment", "UPI Payments"]
spending = [5000, 12000, 8000, 6000]  # example values in INR

# Explode highest spending category
max_index = spending.index(max(spending))
explode = [0.1 if i == max_index else 0 for i in range(len(spending))]

#Create pie chart
fig, ax = plt.subplots()
ax.pie(spending, labels=categories, autopct="%1.1f%%", 
       explode=explode, startangle=90, 
       colors=["orange", "blue", "green", "purple"])

# Add white circle in the center (donut effect)
centre_circle = plt.Circle((0,0), 0.70, fc="white")
ax.add_artist(centre_circle)

ax.set_title("Monthly Online Spending Distribution (Donut Chart)")

plt.show()
```

```python
#4.Suppose you want to compare the market share of 4 IPL cricket teams based on the number of Instagram followers. Build a donut chart where each segment shows 
#the team name and follower 
#percentage, and the chart title clearly states 'IPL Teams Instagram Followers Share'.<br><br><em><strong>Constraint:</strong> 
#Use both labels and autopct to display information on the chart.</em>

teams = ["Mumbai Indians", "Chennai Super Kings", "Royal Challengers Bangalore", "Kolkata Knight Riders"]
followers = [12, 10, 9, 7]  # in millions (example data)

# Create pie chart
fig, ax = plt.subplots()
ax.pie(followers, labels=teams, autopct="%1.1f%%", 
       startangle=90, colors=["blue", "yellow", "red", "purple"])

# Add donut effect (white circle in center)
centre_circle = plt.Circle((0,0), 0.70, fc="white")
ax.add_artist(centre_circle)

ax.set_title("IPL Teams Instagram Followers Share")

plt.show()
```
