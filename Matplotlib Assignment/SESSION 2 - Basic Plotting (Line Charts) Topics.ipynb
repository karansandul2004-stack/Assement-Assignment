```python
#1.Use plt.plot() to create a line chart showing the number of daily steps you walked over the last 7 days. Set the line color to green and linewidth to 2.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4500, 6700, 8000, 7500, 9000, 12000, 11000]

# Create line chart with custom style
plt.plot(days, steps, color="green", linewidth=2, marker="o")

# Add labels and title
plt.title("Daily Steps Walked (Past 7 Days)")
plt.xlabel("Day of Week")
plt.ylabel("Number of Steps")

plt.show()
```

```python
#2.Plot a line chart of your favorite music app's monthly active users for 6 months using plt.plot(). Add circle markers at each data point and set the 
#marker color to red.<br><br><em><strong>Hint:</strong> Use the marker and markerfacecolor parameters in plt.plot().</em>

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
active_users = [120000, 135000, 150000, 160000, 175000, 190000]  # made-up data

# Create line chart with circle markers
plt.plot(months, active_users, 
         color="blue", linewidth=2, 
         marker="o", markerfacecolor="red")

# Add labels and title
plt.title("Monthly Active Users of My Favorite Music App")
plt.xlabel("Month")
plt.ylabel("Active Users")

plt.show()
```

```python
#3.Create a line chart comparing the number of orders you 
#placed on Zomato and Swiggy over the last 5 months. Use different line styles and colors for each app, and add a legend to distinguish them.

months = ["May", "Jun", "Jul", "Aug", "Sep"]
zomato_orders = [25, 30, 28, 35, 40]
swiggy_orders = [20, 27, 32, 30, 38]

# Plot both lines with different styles
plt.plot(months, zomato_orders, 
         color="red", linestyle="-", marker="o", label="Zomato")
plt.plot(months, swiggy_orders, 
         color="blue", linestyle="--", marker="s", label="Swiggy")

# Add labels, title, legend
plt.title("Monthly Orders Comparison: Zomato vs Swiggy")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.legend()

plt.show()
```

```python
#4.Add a main title 'Monthly Data Usage Trend' and label the X-axis as 'Month' and Y-axis as 'Data Used (GB)' to a line 
#chart showing your mobile data usage for the past 6 months.<br><br><em><strong>Hint:</strong> Use plt.title(), plt.xlabel(), and plt.ylabel().</em>

months = ["Apr", "May", "Jun", "Jul", "Aug", "Sep"]
data_usage = [12, 15, 18, 14, 20, 22]  # GB used per month (example data)

#Create line chart
plt.plot(months, data_usage, color="purple", linewidth=2, marker="o")

#Add main title and axis labels
plt.title("Monthly Data Usage Trend")
plt.xlabel("Month")
plt.ylabel("Data Used (GB)")

plt.show()
```
