```python
#1.Create a simple bar chart showing 
#the number of orders for 5 different Zomato restaurants and save the chart as a PNG file using plt.savefig() with a DPI of 150.

import matplotlib.pyplot as plt

restaurants = ["Restaurant A", "Restaurant B", "Restaurant C", "Restaurant D", "Restaurant E"]
orders = [120, 150, 100, 180, 140]

# Create bar chart
plt.figure(figsize=(8,5))
plt.bar(restaurants, orders, color="tomato", edgecolor="black")

# Add labels and title
plt.title("Zomato Orders by Restaurant")
plt.xlabel("Restaurant")
plt.ylabel("Number of Orders")

# Save chart as PNG with DPI=150
plt.savefig("zomato_orders.png", dpi=150)

plt.show()
```

```python
#2.Generate a line plot that displays your daily step count for one week (7 days) and export it as an SVG file using plt.savefig().
#<br><br><em><strong>Hint:</strong> Change the file extension in the filename argument to save as SVG.</em>

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4500, 6700, 8000, 7500, 9000, 12000, 11000]

# Create line plot
plt.figure(figsize=(8,5))
plt.plot(days, steps, marker="o", color="green", linewidth=2)

# Add labels and title
plt.title("Daily Step Count (1 Week)")
plt.xlabel("Day of Week")
plt.ylabel("Steps")

# Save chart as SVG
plt.savefig("weekly_steps.svg", format="svg")

plt.show()
```

```python
#3.Build a pie chart showing the percentage of time you spend on 4 different apps 
#(Instagram, YouTube, WhatsApp, Spotify) in a day and save the chart as a PDF using plt.savefig().

apps = ["Instagram", "YouTube", "WhatsApp", "Spotify"]
time_spent = [120, 90, 60, 100]  # minutes spent (example data)

# Create pie chart
plt.figure(figsize=(7,7))
plt.pie(time_spent, labels=apps, autopct="%1.1f%%", 
        colors=["#F58529", "#FF0000", "#25D366", "#1DB954"], startangle=90)

plt.title("Daily Time Spent on Apps")

# Save chart as PDF
plt.savefig("daily_app_usage.pdf", format="pdf")

plt.show()
```
