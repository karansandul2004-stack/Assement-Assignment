```python
#1.Create a line chart using Matplotlib to show the number of daily 
#active users for a music streaming app (like Spotify) over 12 months, and use plt.annotate() to label the month with the highest user count.

import matplotlib.pyplot as plt

# Prepare data (12 months of active users)
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", 
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
active_users = [1200, 1350, 1500, 1600, 1700, 2000, 
                2100, 2500, 2300, 2400, 2600, 2800]

# Create line chart
plt.figure(figsize=(10,5))
plt.plot(months, active_users, marker="o", color="blue", linewidth=2)

# Find highest user count
max_users = max(active_users)
max_index = active_users.index(max_users)
max_month = months[max_index]

# Annotate highest point
plt.annotate(f"Highest: {max_users}", 
             xy=(max_month, max_users), 
             xytext=(max_index, max_users+200),
             arrowprops=dict(facecolor="red", shrink=0.05),
             ha="center")

# Add labels and title
plt.title("Monthly Active Users on Music Streaming App")
plt.xlabel("Month")
plt.ylabel("Active Users")

plt.show()
```

```python
#2.Plot a bar chart showing the number of orders per month for a food delivery app (like Zomato), and add an arrow annotation using plt.annotate() 
#to highlight the month with 
#the lowest orders.<br><br><em><strong>Hint:</strong> Use the 'arrowprops' parameter to draw the arrow pointing to the bar.</em>

# Prepare data (12 months of orders)
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", 
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
orders = [120, 150, 100, 180, 170, 200, 190, 220, 210, 230, 250, 240]

# Create bar chart
plt.figure(figsize=(10,5))
bars = plt.bar(months, orders, color="orange")

# Find lowest orders
min_orders = min(orders)
min_index = orders.index(min_orders)
min_month = months[min_index]

# Annotate lowest bar
plt.annotate(f"Lowest: {min_orders}", 
             xy=(min_index, min_orders), 
             xytext=(min_index, min_orders+50),
             arrowprops=dict(facecolor="blue", shrink=0.05),
             ha="center")

# Add labels and title
plt.title("Monthly Food Orders on Zomato")
plt.xlabel("Month")
plt.ylabel("Number of Orders")

plt.show()
```

```python
#3.Create a pie chart representing the percentage of different payment methods 
#(UPI, Card, Wallet, Cash) used on an e-commerce app (like Flipkart), and add a legend positioned at the lower right corner of the plot.

payment_methods = ["UPI", "Card", "Wallet", "Cash"]
usage = [45, 30, 15, 10]  # example percentages

# Create pie chart
plt.figure(figsize=(7,7))
plt.pie(usage, labels=payment_methods, autopct="%1.1f%%", 
        colors=["#FF5733", "#33C1FF", "#9D33FF", "#33FF57"], startangle=90)

# Add legend at lower right
plt.legend(payment_methods, loc="lower left")

# Add title
plt.title("Flipkart Payment Methods Distribution")

plt.show()
```

```python
#4.Given a scatter plot of movie ratings vs. number of reviews for movies on a streaming platform (like BookMyShow), use plt.annotate() to add a 
#text box highlighting the 
#movie with the highest rating.<br><br><em><strong>Hint:</strong> Set the 'bbox' parameter in plt.annotate() to create a visible text box.</em>

movies = ["Movie A", "Movie B", "Movie C", "Movie D", "Movie E"]
ratings = [7.8, 8.5, 9.2, 8.0, 7.5]       # Ratings out of 10
reviews = [120, 300, 150, 400, 250]       # Number of reviews

# Create scatter plot
plt.figure(figsize=(8,6))
plt.scatter(reviews, ratings, color="purple", marker="o")

# Find highest rating
max_rating = max(ratings)
max_index = ratings.index(max_rating)
highlight_movie = movies[max_index]

# Annotate with text box
plt.annotate(f"Highest Rated: {highlight_movie}\nRating: {max_rating}",
             xy=(reviews[max_index], ratings[max_index]),
             xytext=(reviews[max_index]+50, ratings[max_index]-0.3),
             bbox=dict(boxstyle="round,pad=0.3", fc="yellow", alpha=0.7),
             arrowprops=dict(facecolor="black", arrowstyle="->"),
             fontsize=10)

# Add labels and title
plt.title("Movie Ratings vs. Number of Reviews")
plt.xlabel("Number of Reviews")
plt.ylabel("Rating")

plt.show()
```
