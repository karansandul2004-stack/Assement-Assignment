```python
#2.Create a scatter plot showing 'amount spent on Zomato orders' vs 'number of orders placed' for 12 
#friends, and label each point with the friend's name.<br><br><em><strong>Hint:</strong> Use the 'annotate' function in matplotlib to add labels.</em>

import matplotlib.pyplot as plt

# Prepare fictional data
friends = ["Aman", "Riya", "Karan", "Neha", "Arjun", "Simran", 
           "Rahul", "Priya", "Vikram", "Sneha", "Dev", "Meera"]

amount_spent = [1200, 950, 1800, 1100, 2000, 750, 
                1600, 1300, 2200, 900, 1400, 1700]

orders_placed = [15, 12, 20, 14, 22, 10, 
                 18, 16, 25, 11, 17, 19]

# Create scatter plot
plt.scatter(amount_spent, orders_placed, color="red", marker="o")

# Add labels and title
plt.title("Zomato Orders: Amount Spent vs Number of Orders")
plt.xlabel("Amount Spent (INR)")
plt.ylabel("Number of Orders Placed")

# Annotate each point with friend's name
for i, name in enumerate(friends):
    plt.annotate(name, (amount_spent[i], orders_placed[i]), 
                 textcoords="offset points", xytext=(5,5), ha="left")

plt.show()
```

```python
#3.Add a trend line to your Zomato scatter plot by calculating the best-fit line using numpy's polyfit(), 
#then plot it on top of the scatter plot.<br><br><em><strong>Hint:</strong> Use np.polyfit() to get the slope and intercept, then plot the line with plt.plot().</em>

import matplotlib.pyplot as plt
import numpy as np

# Prepare fictional data
friends = ["Aman", "Riya", "Karan", "Neha", "Arjun", "Simran", 
           "Rahul", "Priya", "Vikram", "Sneha", "Dev", "Meera"]

amount_spent = [1200, 950, 1800, 1100, 2000, 750, 
                1600, 1300, 2200, 900, 1400, 1700]

orders_placed = [15, 12, 20, 14, 22, 10, 
                 18, 16, 25, 11, 17, 19]

# Scatter plot
plt.scatter(amount_spent, orders_placed, color="red", marker="o")

# Annotate each point with friend's name
for i, name in enumerate(friends):
    plt.annotate(name, (amount_spent[i], orders_placed[i]), 
                 textcoords="offset points", xytext=(5,5), ha="left")

# Calculate best-fit line using polyfit
slope, intercept = np.polyfit(amount_spent, orders_placed, 1)
trend_line = [slope*x + intercept for x in amount_spent]

# Plot trend line
plt.plot(amount_spent, trend_line, color="blue", linewidth=2, label="Trend Line")

# Add labels and title
plt.title("Zomato Orders: Amount Spent vs Number of Orders")
plt.xlabel("Amount Spent (INR)")
plt.ylabel("Number of Orders Placed")
plt.legend()

plt.show()
```

```python
#4.Given a dataset of 'YouTube watch hours' and 'Spotify 
#listening hours' for 15 users, use numpy polyfit to fit a regression line, plot both the scatter and the trend line, and print the equation of the line on the plot.

import numpy as np

# Sample data (15 users)
youtube_hours = [2.5, 3.0, 4.5, 5.0, 3.2, 2.0, 4.0, 6.0, 3.8, 4.2, 5.5, 2.8, 3.6, 4.8, 5.2]
spotify_hours = [1.5, 2.0, 3.5, 4.0, 2.2, 1.0, 3.0, 5.0, 2.8, 3.2, 4.5, 1.8, 2.6, 3.8, 4.2]

# Scatter plot
plt.scatter(youtube_hours, spotify_hours, color="purple", marker="o", label="Users")

# Fit regression line using polyfit
slope, intercept = np.polyfit(youtube_hours, spotify_hours, 1)
trend_line = [slope*x + intercept for x in youtube_hours]

# Plot regression line
plt.plot(youtube_hours, trend_line, color="blue", linewidth=2, label="Trend Line")

# Add equation of line on plot
equation = f"y = {slope:.2f}x + {intercept:.2f}"
plt.text(min(youtube_hours)+0.5, max(spotify_hours)-0.5, equation, fontsize=10, color="blue")

# Add labels and title
plt.title("YouTube Watch Hours vs Spotify Listening Hours")
plt.xlabel("YouTube Watch Hours (per day)")
plt.ylabel("Spotify Listening Hours (per day)")
plt.legend()

plt.show()
```
