```python
# hello_plot.py
import matplotlib.pyplot as plt

# Step 1: Prepare data (daily steps for 7 days)
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4500, 6700, 8000, 7500, 9000, 12000, 11000]

# Step 2: Create plot
plt.figure(figsize=(8, 5))  # Create figure
plt.plot(days, steps, marker='o', linestyle='-', color='blue')

# Step 3: Add labels and title
plt.title("Daily Steps Walked (Past 7 Days)")
plt.xlabel("Day of Week")
plt.ylabel("Number of Steps")

# Step 4: Show plot
plt.show()
```

```python
#2.Create a script called insta_likes_plot.py that uses Matplotlib to plot 
#the number of likes you received on your last 10 Instagram posts (use made-up data if needed), labeling the axes appropriately.

# insta_likes_plot.py
import matplotlib.pyplot as plt

# Prepare data (likes on 10 posts)
posts = [f"Post {i}" for i in range(1, 11)]
likes = [120, 340, 560, 430, 800, 650, 720, 500, 900, 1000]

# Create plot
plt.figure(figsize=(9, 5))
plt.plot(posts, likes, marker='o', linestyle='-', color='purple')

# Add labels and title
plt.title("Instagram Likes on Last 10 Posts")
plt.xlabel("Posts")
plt.ylabel("Number of Likes")

plt.show()
```

```python
#3.Write a Python script that creates a Matplotlib figure with two Axes: one showing your daily Spotify listening 
#hours for a week and the other showing your daily YouTube viewing hours for the same week. Display both plots in a single figure.

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
spotify_hours = [2.5, 3.0, 1.5, 2.0, 4.0, 5.0, 3.5]
youtube_hours = [1.0, 2.5, 3.0, 2.0, 3.5, 4.5, 2.5]

# Create figure with two Axes (subplots)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Plot Spotify data
ax1.plot(days, spotify_hours, marker='o', color='green', linestyle='-')
ax1.set_title("Daily Spotify Listening Hours")
ax1.set_xlabel("Day of Week")
ax1.set_ylabel("Hours")

# Plot YouTube data
ax2.plot(days, youtube_hours, marker='s', color='red', linestyle='--')
ax2.set_title("Daily YouTube Viewing Hours")
ax2.set_xlabel("Day of Week")
ax2.set_ylabel("Hours")

# Adjust layout and show
plt.tight_layout()
plt.show()
```
