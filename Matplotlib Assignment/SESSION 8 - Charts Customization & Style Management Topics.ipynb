```python
#1.Create a line chart showing the number of daily active users 
#on a music streaming app (like Spotify) over 7 days, and set the figure size to 10 inches wide by 4 inches tall using plt.figure().

import matplotlib.pyplot as plt

# Prepare data (7 days of daily active users)
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
active_users = [1200, 1350, 1500, 1600, 1700, 2000, 1800]

# Set figure size
plt.figure(figsize=(10,4))

# Create line chart
plt.plot(days, active_users, marker="o", linestyle="-", color="purple", linewidth=2)

# Add labels and title
plt.title("Daily Active Users on Music Streaming App (7 Days)")
plt.xlabel("Day of Week")
plt.ylabel("Active Users")

plt.show()
```

```python
#2.Plot two lines on the same graph representing the number of food orders on 
#Zomato and Swiggy over a week, using different line styles (e.g., dashed for Zomato, dotted for Swiggy) and add a legend to distinguish them.

# Prepare data (7 days of orders)
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
zomato_orders = [12, 15, 14, 18, 20, 22, 19]
swiggy_orders = [10, 13, 12, 16, 18, 21, 17]

# Create line chart with different styles
plt.plot(days, zomato_orders, linestyle="--", color="red", marker="o", label="Zomato")
plt.plot(days, swiggy_orders, linestyle=":", color="blue", marker="s", label="Swiggy")

# Add labels, title, and legend
plt.title("Weekly Food Orders: Zomato vs Swiggy")
plt.xlabel("Day of Week")
plt.ylabel("Number of Orders")
plt.legend()

plt.show()
```

```python
#3.Apply the 'ggplot' style to a chart showing Flipkart's daily sales for 10 days using plt.style.use('ggplot'), and customize the line color to orange.

# Apply ggplot style
plt.style.use("ggplot")

# Prepare data (10 days of sales)
days = list(range(1, 11))  # Day 1 to Day 10
sales = [1200, 1500, 1700, 1600, 1800, 2000, 2200, 2100, 2300, 2500]

# Create line chart with custom color
plt.figure(figsize=(8,4))
plt.plot(days, sales, color="orange", marker="o", linewidth=2)

# Add labels and title
plt.title("Flipkart Daily Sales (10 Days)")
plt.xlabel("Day")
plt.ylabel("Sales (INR)")

plt.show()
```

```python
#4.Choose a color palette that matches Instagram's branding and plot a bar chart showing the number of posts uploaded each day for a week; 
#use those colors for the bars.<br><br><em><strong>Hint:</strong> You can use a list of hex color codes to set bar colors.</em>

# Prepare data (posts per day for a week)
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
posts = [5, 7, 6, 8, 10, 12, 9]

# Instagram-inspired color palette (hex codes)
insta_colors = ["#F58529", "#DD2A7B", "#8134AF", "#515BD4", "#FFDC80", "#FD1D1D", "#C13584"]

# Create bar chart
plt.bar(days, posts, color=insta_colors, edgecolor="black")

# Add labels and title
plt.title("Instagram Posts Uploaded per Day (1 Week)")
plt.xlabel("Day of Week")
plt.ylabel("Number of Posts")

plt.show()
```

```python
#5.Take any one of your previous charts and restyle it to match a dark theme using plt.style.use('dark_background'), then adjust the line or 
#bar colors so that all text and 
#data remain clearly visible.<br><br><em><strong>Constraint:</strong> Do not use the same colors as in the original chart.</em>

# Apply dark theme
plt.style.use("dark_background")

# Prepare data (10 days of sales)
days = list(range(1, 11))  # Day 1 to Day 10
sales = [1200, 1500, 1700, 1600, 1800, 2000, 2200, 2100, 2300, 2500]

# Create line chart with new colors (not orange)
plt.figure(figsize=(8,4))
plt.plot(days, sales, color="cyan", marker="o", linewidth=2)

# Add labels and title
plt.title("Flipkart Daily Sales (10 Days) - Dark Theme")
plt.xlabel("Day")
plt.ylabel("Sales (INR)")

plt.show()
```
