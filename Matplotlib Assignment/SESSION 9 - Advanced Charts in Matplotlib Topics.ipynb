```python
#1.Create a stacked bar chart using Matplotlib that shows 
#the number of food orders from Zomato and Swiggy for each day of the week. Use different colors for each platform and add a legend.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
zomato_orders = [12, 15, 14, 18, 20, 22, 19]
swiggy_orders = [10, 13, 12, 16, 18, 21, 17]

# Create stacked bar chart
plt.bar(days, zomato_orders, color="red", label="Zomato")
plt.bar(days, swiggy_orders, bottom=zomato_orders, color="blue", label="Swiggy")

# Add labels, title, and legend
plt.title("Weekly Food Orders: Zomato vs Swiggy (Stacked)")
plt.xlabel("Day of Week")
plt.ylabel("Number of Orders")
plt.legend()

plt.show()
```

```python
#2.Plot an area chart that visualizes the daily active users on Spotify and YouTube over a 7-day period using 
#Matplotlib's fill_between(). Make sure each platform's area is filled with a different color and add labels for clarity.

# Prepare data (7 days)
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
spotify_users = [1200, 1350, 1500, 1600, 1700, 2000, 1800]
youtube_users = [1500, 1600, 1700, 1900, 2100, 2500, 2200]

# Create area chart
plt.figure(figsize=(10,5))

# Spotify area
plt.fill_between(days, spotify_users, color="green", alpha=0.5, label="Spotify")

# YouTube area
plt.fill_between(days, youtube_users, color="red", alpha=0.5, label="YouTube")

# Add labels and title
plt.title("Daily Active Users: Spotify vs YouTube (7 Days)")
plt.xlabel("Day of Week")
plt.ylabel("Active Users")
plt.legend()

plt.show()
```

```python
#3.Simulate the average delivery time (in minutes) for Flipkart orders over 10 days, and add error bars to represent the standard deviation 
#for each day's delivery time using 
#Matplotlib's errorbar() function.<br><br><em><strong>Hint:</strong> Use the yerr parameter to show error bars for each data point.</em>

import numpy as np

# Simulated data (10 days)
days = list(range(1, 11))  # Day 1 to Day 10
avg_delivery_time = [32, 30, 28, 35, 33, 31, 29, 34, 36, 32]  # average minutes
std_dev = [3, 2, 4, 3, 2, 3, 2, 4, 3, 2]  # standard deviation

# Plot with error bars
plt.figure(figsize=(8,5))
plt.errorbar(days, avg_delivery_time, yerr=std_dev, fmt="o-", 
             color="blue", ecolor="red", capsize=5, label="Avg Delivery Time")

# Add labels and title
plt.title("Flipkart Orders: Average Delivery Time with Error Bars")
plt.xlabel("Day")
plt.ylabel("Delivery Time (minutes)")
plt.legend()

plt.show()
```

```python
#4.Plot a time series line chart showing your daily 
#Paytm wallet balance over the last 14 days using Matplotlib. Format the x-axis to display dates in 'DD MMM' format (e.g., '12 Jun').

import matplotlib.dates as mdates
import datetime as dt

# Generate dates (last 14 days)
dates = [dt.date(2026, 9, 14) + dt.timedelta(days=i) for i in range(14)]

# Simulated Paytm wallet balances (in INR)
balances = [1500, 1600, 1550, 1700, 1650, 1800, 1750,
            1900, 1850, 2000, 1950, 2100, 2050, 2200]

# Create time series line chart
plt.figure(figsize=(10,5))
plt.plot(dates, balances, marker="o", color="teal", linewidth=2)

# Format x-axis to show 'DD MMM'
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%d %b"))

# Add labels and title
plt.title("Daily Paytm Wallet Balance (Last 14 Days)")
plt.xlabel("Date")
plt.ylabel("Balance (INR)")

# Rotate date labels for readability
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
